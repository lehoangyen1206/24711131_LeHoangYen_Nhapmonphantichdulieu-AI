# PHẦN B — DESCRIPTIVE STATISTICS
## Group 1 — Central Tendency

```python
numeric_cols = ['age', 'fare', 'sibsp', 'parch']
for col in numeric_cols:
    print(f"\n--- {col} ---")
    print("Mean:", df[col].mean())
    print("Median:", df[col].median())
    print("Mode:", df[col].mode().iloc[0])
```

## Group 2 — Dispersion

```python
for col in numeric_cols:
    print(f"\n--- {col} ---")
    print("Range:", df[col].max() - df[col].min())
    print("Variance:", df[col].var())
    print("Standard deviation:", df[col].std())
    print("IQR:", df[col].quantile(0.75) - df[col].quantile(0.25))
```

## Group 3 — Location and Shape

```python
for col in numeric_cols:
    print(f"\n--- {col} ---")
    print("Min:", df[col].min())
    print("Q1:", df[col].quantile(0.25))
    print("Median:", df[col].median())
    print("Q3:", df[col].quantile(0.75))
    print("Max:", df[col].max())
    print("Skewness:", df[col].skew())
    print("Kurtosis:", df[col].kurt())
```

# PHẦN C — DEFINE THE QUESTION

## Câu hỏi 1: Hạng vé nào có tỷ lệ sống sót cao nhất, chênh lệch bao nhiêu so với hạng thấp nhất?

```python
hang_ve_song_sot = df.groupby('class')['survived'].mean().sort_values(ascending=False)

print("Tỷ lệ sống sót theo hạng vé:")
print(hang_ve_song_sot)

print(f"\nHạng vé có tỷ lệ sống sót cao nhất: {hang_ve_song_sot.index[0]}")
print(f"Chênh lệch với hạng thấp nhất: {(hang_ve_song_sot.iloc[0] - hang_ve_song_sot.iloc[-1]) * 100:.2f}%")
```

## Câu hỏi 2: Giới tính hay hạng vé ảnh hưởng đến sống sót mạnh hơn?

```python
gioi_tinh = df.groupby('sex')['survived'].mean()
hang_ve = df.groupby('class')['survived'].mean()

gioi_tinh1 = gioi_tinh.max() - gioi_tinh.min()
hang_ve1 = hang_ve.max() - hang_ve.min()

print("Tỷ lệ sống sót theo giới tính:")
print(gioi_tinh)

print("\nTỷ lệ sống sót theo hạng vé:")
print(hang_ve)

print(f"\nChênh lệch theo giới tính: {gioi_tinh1 * 100:.2f}%")
print(f"Chênh lệch theo hạng vé: {hang_ve1 * 100:.2f}%")

if gioi_tinh1 > hang_ve1:
    print("Giới tính ảnh hưởng mạnh hơn đến khả năng sống sót")
else:
    print("Hạng vé ảnh hưởng mạnh hơn đến khả năng sống sót")
```

## Câu hỏi 3: Vé đắt hơn có thực sự sống sót cao hơn không?

```python
tg_quan = df[['fare', 'survived']].corr().loc['fare', 'survived']

print(f"Tương quan giữa giá vé và sống sót: {tg_quan:.4f}")

df['fare_group'] = pd.qcut(df['fare'], 4, labels=['Thấp', 'Trung bình thấp', 'Trung bình cao', 'Cao'])
ve_dat = df.groupby('fare_group', observed=False)['survived'].mean()
print("\nTỷ lệ sống sót theo nhóm giá vé:")
print(ve_dat)

if tg_quan > 0:
    print("\nGiá vé có tương quan dương với khả năng sống sót")
else:
    print("\nGiá vé không có tương quan dương với khả năng sống sót")
```

## Câu hỏi 4: Gia đình đông người có ảnh hưởng đến khả năng sống sót không?

```python
gd = df.groupby('family_size')['survived'].mean()

print("Tỷ lệ sống sót theo quy mô gia đình:")
print(gd)

tg_quan_gd = df[['family_size', 'survived']].corr().loc['family_size', 'survived']

print(f"\nTương quan giữa quy mô gia đình và khả năng sống sót: {tg_quan_gd:.4f}")
```

## Câu hỏi 5: Cảng lên tàu (embark_town) nào có tỷ lệ sống sót cao nhất?

```python
cang = df.groupby('embark_town')['survived'].mean().sort_values(ascending=False)

print("Tỷ lệ sống sót theo cảng:")
print(cang)

print(f"\nCảng có tỷ lệ sống sót cao nhất: {cang.index[0]}")
print(f"Tỷ lệ: {cang.iloc[0] * 100:.2f}%")
```
