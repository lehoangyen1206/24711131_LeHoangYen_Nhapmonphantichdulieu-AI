## Setup

```python
import pandas as pd, numpy as np, matplotlib.pyplot as plt, seaborn as sns
from pathlib import Path
from scipy import stats

sns.set_style('whitegrid')


csv_path = 'https://raw.githubusercontent.com/stedy/Machine-Learning-with-R-datasets/master/insurance.csv'

df = pd.read_csv(csv_path)
print('Loaded from:', csv_path)
df.head()
```

---
# PHẦN A — DATA PROFILING
## A.1. Data size, column names, data types

```python
df.info()
```

## A.2. Missing values & Duplicate data

```python
print("Kiểm tra dữ liệu bị thiếu")
print(df.isnull().sum())

print("\nKiểm tra trùng lặp")
print(df.duplicated().sum())
```

## A.3. Invalid values

```python
print("Age không hợp lệ:")
print(df[df['age'] <= 0])

print("\nBMI không hợp lệ:")
print(df[df['bmi'] <= 0])

print("\nChildren không hợp lệ:")
print(df[df['children'] < 0])

print("\nCharges không hợp lệ:")
print(df[df['charges'] <= 0])

print("\nSex không hợp lệ:")
print(df[~df['sex'].isin(['male', 'female'])])

print("\nSmoker không hợp lệ:")
print(df[~df['smoker'].isin(['yes', 'no'])])

print("\nRegion không hợp lệ:")
print(df[~df['region'].isin(['southwest', 'southeast', 'northwest', 'northeast'])])
```

## A.4. Create a new column
Tạo cột `bmi_group`: Normal (<25), Overweight (25-30), Obese (>=30).

```python
df['bmi_group'] = df['bmi'].apply(
    lambda x: 'Normal' if x < 25 else ('Obese' if x >= 30 else 'Overweight')
)
df[['bmi', 'bmi_group']]
```

# PHẦN B — DESCRIPTIVE STATISTICS
## Group 1 — Central Tendency

```python
numeric_cols = ['age', 'bmi', 'children', 'charges']

for col in numeric_cols:
    print(f"\n--- {col} ---")
    print("Mean:", df[col].mean())
    print("Median:", df[col].median())
    print("Mode:", df[col].mode().tolist())
```

## Group 2 — Dispersion

```python
for col in numeric_cols:
    print(f"\n--- {col} ---")
    
    print("Range:", df[col].max() - df[col].min())
    print("Variance:", df[col].var())
    print("Standard deviation:", df[col].std())
    print("IQR:", df[col].quantile(0.75) - df[col].quantile(0.25))
```

## Group 3 — Location and Shape

```python
for col in numeric_cols:
    print(f"\n--- {col} ---")
    
    print("Min:", df[col].min())
    print("Q1:", df[col].quantile(0.25))
    print("Median:", df[col].median())
    print("Q3:", df[col].quantile(0.75))
    print("Max:", df[col].max())
    print("Skewness:", df[col].skew())
    print("Kurtosis:", df[col].kurt())
```

# PHẦN C — DEFINE THE QUESTION

## Câu hỏi 1: Người hút thuốc trả chi phí cao hơn bao nhiêu lần so với người không hút, và có đồng đều giữa các vùng không?

```python
TB = df.groupby('smoker')['charges'].mean()
print(TB)

ty_le = TB['yes'] / TB['no']
print(f"Người hút thuốc lá phải trả cao hơn khoảng {ty_le:.2f} lần")
```

```python
vung = df.groupby(['region', 'smoker'])['charges'].mean().unstack()
print(vung)

vung['ty_le'] = vung['yes'] / vung['no']
print("\nTỷ lệ người hút / người không hút theo vùng:")
print(vung['ty_le'].sort_values(ascending=False))
```

## Câu hỏi 2: BMI có tương quan với chi phí mạnh hơn ở nhóm hút thuốc hay không hút thuốc?

```python
ng_hut = df[df['smoker'] == 'yes'][['bmi', 'charges']].corr().loc['bmi', 'charges']
k_hut = df[df['smoker'] == 'no'][['bmi', 'charges']].corr().loc['bmi', 'charges']

print(f"Người hút thuốc: {ng_hut:.4f}")
print(f"Người không hút thuốc: {k_hut:.4f}")
if abs(ng_hut) > abs(k_hut):
    print("BMI có tương quan mạnh hơn với chi phí ở nhóm hút thuốc")
else:
    print("BMI có tương quan mạnh hơn với chi phí ở nhóm không hút thuốc")
```

## Câu hỏi 3: Vùng nào có chi phí bảo hiểm trung bình cao nhất?

```python
tb_vung = df.groupby('region')['charges'].mean().sort_values(ascending = False)
print(tb_vung)
print(f"Vùng có chi phí bảo hiểm trung bình cao nhất: {tb_vung.index[0]}")
```

## Câu hỏi 4: Số lượng con cái có làm tăng chi phí bảo hiểm không?

```python
tb_concai = df.groupby('children')['charges'].mean()
print(tb_concai)
tg_quan_con = df[['children', 'charges']].corr().loc['children', 'charges']
print(f"\nTương quan giữa số con và chi phí: {tg_quan_con:.4f}")
```

## Câu hỏi 5: Tuổi có tương quan với chi phí bảo hiểm không?

```python
tg_quan_tuoi = df[['age', 'charges']].corr().loc['age', 'charges']
print(f"\nTương quan giữa tuổi và chi phí: {tg_quan_tuoi:.4f}")
```
